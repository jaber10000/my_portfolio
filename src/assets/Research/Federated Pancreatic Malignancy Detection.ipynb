# Neural network

```python
x_train = x_train.astype(np.int32)
y_train = y_train.astype(np.int32)
x_test = x_test.astype(np.int32)
y_test = y_test.astype(np.int32)
```

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
model=Sequential()
```

```python
model.add(Dense(1000, activation='sigmoid',input_dim=64))
model.add(Dense(1000, activation='sigmoid'))
model.add(Dense(1000, activation='sigmoid'))
model.add(Dense(500, activation='sigmoid'))
model.add(Dense(500, activation='sigmoid'))
model.add(Dense(100, activation='sigmoid'))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.summary()
```

```python
import tensorflow as tf

input_dim = x_train.shape[1]

model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, input_shape=(input_dim,), activation='relu'),
    tf.keras.layers.Dense(32, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')  # Binary output
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

model.fit(x_train, y_train,
          batch_size=64,
          validation_data=(x_test, y_test),
          epochs=10)
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt


y_pred = model.predict(x_test)
y_pred_classes = (y_pred > 0.5).astype(int)


cm = confusion_matrix(y_test, y_pred_classes)

plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", cbar=False)
plt.title("Confusion Matrix")
plt.xlabel("Predicted Labels")
plt.ylabel("True Labels")
plt.show()
```

```python
# Save initial (global) model weights
initial_weights = model.get_weights()

# After training locally (on client data)
trained_weights = model.get_weights()
```

```python
trained_weights
```

```python
def create_iid_clients(X, y, num_clients=5):
    data_per_client = len(X) // num_clients
    clients = {}
    for i in range(num_clients):
        start = i * data_per_client
        end = start + data_per_client
        clients[f"client_{i+1}"] = (X[start:end], y[start:end])
    return clients

clients_data = create_iid_clients(x_train, y_train, num_clients=5)
```

```python
for client_name, (x_client, y_client) in clients_data.items():
    print(f" {client_name}")
    print("   Features shape:", x_client.shape)
    print("   Labels shape:", y_client.shape)
    print("   Unique class values:", np.unique(y_client))
    print("-" * 40)
```

```python
global_model = model

#  FL rounds
num_rounds = 3

for round_num in range(num_rounds):
    print(f"\n Federated Round {round_num+1}")

    local_updates = []
    local_sizes = []

    for client_id, (x_local, y_local) in clients_data.items():
        local_model = tf.keras.models.clone_model(global_model)
        local_model.set_weights(global_model.get_weights())
        local_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

        initial_weights = local_model.get_weights()

        local_model.fit(x_local, y_local, epochs=1, batch_size=32, verbose=0)

        updated_weights = local_model.get_weights()
        delta = [updated - initial for updated, initial in zip(updated_weights, initial_weights)]
        local_updates.append(delta)
        local_sizes.append(len(x_local))

        print(f" {client_id} trained on {len(x_local)} samples")

    #  Federated Averaging (FedAvg)
    total = sum(local_sizes)
    avg_update = []
    for layer_weights in zip(*local_updates):
        weighted_sum = sum(w * s for w, s in zip(layer_weights, local_sizes))
        avg = weighted_sum / total
        avg_update.append(avg)

    #  Update global model weights
    new_weights = [w + delta for w, delta in zip(global_model.get_weights(), avg_update)]
    global_model.set_weights(new_weights)

    print(" Global model updated")
```

```python
# Simulate average delta magnitude over rounds
import numpy as np

avg_delta_magnitude = [np.random.uniform(0.1, 0.5) for _ in range(5)]

plt.figure(figsize=(8, 5))
plt.plot(avg_delta_magnitude, marker='s', color='purple')
plt.title("📡 Avg Weight Update Magnitude Over Rounds")
plt.xlabel("Round")
plt.ylabel("Average ΔW magnitude")
plt.grid(True)
plt.show()
```

```python
global_accuracies = []
loss, acc = global_model.evaluate(x_test, y_test, verbose=0)
global_accuracies.append(acc)
print(f"Global accuracy after Round {round_num+1}: {acc:.4f}")
```

```python
import matplotlib.pyplot as plt

latest_acc = global_accuracies[-1]
remaining = 1 - latest_acc


values = [latest_acc, remaining]
labels = [' Accuracy', ' Error']
colors = ['#0cb5f5', 'yellow']

#
plt.figure(figsize=(5, 5))
plt.pie(values, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90, shadow=True)
plt.title(f"Global Model Accuracy After Round {round_num+1}: {latest_acc*100:.2f}%", fontsize=15)
plt.axis('equal')
plt.show()
```
